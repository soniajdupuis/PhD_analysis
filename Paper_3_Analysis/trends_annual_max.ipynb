```python
import numpy as np
from scipy.stats import theilslopes

def theil_sen_1d(y, x):
    # remove NaNs
    mask = np.isfinite(y) & np.isfinite(x)
    
    # not enough data points → return NaN
    if mask.sum() < 5:
        return np.nan
    
    slope, intercept, lower, upper = theilslopes(y[mask], x[mask])
    return slope
```

```python
slope = xr.apply_ufunc(
    theil_sen_1d,
    anoms,
    time_numeric,
    input_core_dims=[["time"], ["time"]],
    output_core_dims=[[]],
    vectorize=True,
    dask="parallelized",
    output_dtypes=[float],
)

trend_per_decade = slope * 1
```

```python
import pymannkendall as mk

def mk_pvalue(y):
    y = y[np.isfinite(y)]
    if y.size < 5:
        return np.nan
    return mk.original_test(y).p
```

```python
from scipy.stats import theilslopes

def theil_sen_1d(y, x):
    slope, intercept, lower, upper = theilslopes(y, x)
    return slope
```

```python
theil_sen_trend = xr.apply_ufunc(
    theil_sen_1d,
    anom_grouped,
    year_grouped,
    input_core_dims=[["time"], ["time"]],
    output_core_dims=[[]],
    vectorize=True,
    dask="parallelized",
    output_dtypes=[float],
)
```

```python
trend_per_decade = theil_sen_trend * 10
```
